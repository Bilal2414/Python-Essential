```python
array_1 = np.array([5,7,8,9])
array_2 = np.array([3,5,7,9])
print(array_1 + array_2) #Adding two arrays
print(array_1 - array_2) #Subtracting two arrays
print(array_1 * array_2) #Multiplying two arrays
print(array_1 / array_2) #Dividing two arrays
```

```python
print(np.sum(array_1)) #Sum of all elements in the array
```
